```python
df = df.reset_index()

print(df.head())
print(df.info())
```

```python
# 기존 데이터 복사
df_updated = df.copy()

# 숫자 열을 결측치 입력이 가능한 자료형으로 변경
df_updated = df_updated.astype('Int64')

# 1249회 빈 데이터 만들기
new_row = pd.DataFrame([{
    '회차': 1249,
    '번호1': pd.NA,
    '번호2': pd.NA,
    '번호3': pd.NA,
    '번호4': pd.NA,
    '번호5': pd.NA,
    '번호6': pd.NA,
    '보너스': pd.NA
}])

new_row = new_row.astype('Int64')
```

```python
print("전체 데이터:", len(df_updated))
print("최신 회차:", df_updated['회차'].max())

display(df_updated.head(3))

display(df_updated[df_updated['회차'] == 1249])
```

```python
# 1249회 빈 행 생성
new_row = pd.DataFrame([{
    '회차': 1249,
    '번호1': pd.NA,
    '번호2': pd.NA,
    '번호3': pd.NA,
    '번호4': pd.NA,
    '번호5': pd.NA,
    '번호6': pd.NA,
    '보너스': pd.NA
}])

# 기존 데이터와 합치기
df_updated = pd.concat(
    [df, new_row],
    ignore_index=True
)

# 회차 내림차순 정렬
df_updated = df_updated.sort_values(
    '회차',
    ascending=False
).reset_index(drop=True)

# 결과 즉시 출력
print("전체 데이터:", len(df_updated))
print("최신 회차:", df_updated['회차'].max())

display(df_updated.head(3))
display(df_updated[df_updated['회차'] == 1249])
```

```python
# CSV 파일 저장
df_updated.to_csv(
    'lotto_1_1249.csv',
    index=False,
    encoding='utf-8-sig'
)

# 엑셀 파일 저장
df_updated.to_excel(
    'lotto_1_1249.xlsx',
    index=False
)

print("CSV 저장 완료: lotto_1_1249.csv")
print("엑셀 저장 완료: lotto_1_1249.xlsx")
```
